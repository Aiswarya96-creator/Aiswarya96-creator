```python
def get_product_price(product_id, products_list):
    for product in products_list:
        if product["Product_ID"] == product_id:
            return product["Price"]

    return None
print(get_product_price("P08", products))
```

```python
def get_customer_name(customer_id, customers_list):
    for customer in customers_list:
        if customer["Customer_ID"] == customer_id:
            return customer["Name"]

    return None
print(get_customer_name("C04", customers))
```

```python
def calculate_order_value(price, quantity):
    return price * quantity

print(calculate_order_value(799, 3))
```

```python
def classify_order(value, high_limit=3000, low_limit=1000):
    if value > high_limit:
        return "High"
    elif value >= low_limit:
        return "Medium"
    else:
        return "Low"

print(classify_order(3500))
```

```python
def generate_invoice(order, products_list, customers_list):

    product_price = get_product_price(order["Product_ID"], products_list)
    customer_name = get_customer_name(order["Customer_ID"], customers_list)

    order_value = calculate_order_value(
        product_price,
        order["Quantity"]
    )

    invoice = f"""


Order ID: {order["Order_ID"]}
Customer: {customer_name}
Quantity: {order["Quantity"]}
Price: {product_price}
Order Value: {order_value}
Payment Mode: {order["Payment_Mode"]}
Status: {order["Status"]}

"""

    return invoice

print(generate_invoice(orders[0], products, customers))
```

```python
def calculate_total_revenue(orders_list, products_list):

    total = 0

    for order in orders_list:

        if order["Status"] == "Cancelled":
            continue

        price = get_product_price(order["Product_ID"], products_list)

        order_value = calculate_order_value(
            price,
            order["Quantity"]
        )

        total += order_value

    return total


revenue = calculate_total_revenue(orders, products)

print("Total Revenue:", revenue)
```

```python
def top_customer_by_spend(orders_list, products_list, customers_list):

    spending = {}

    for customer in customers_list:
        spending[customer["Customer_ID"]] = 0

    for order in orders_list:

        if order["Status"] != "Delivered":
            continue

        price = get_product_price(order["Product_ID"], products_list)

        order_value = calculate_order_value(
            price,
            order["Quantity"]
        )

        spending[order["Customer_ID"]] += order_value

    top_customer_id = None
    highest_spend = 0

    for customer_id in spending:
        if spending[customer_id] > highest_spend:
            highest_spend = spending[customer_id]
            top_customer_id = customer_id

    return get_customer_name(top_customer_id, customers_list)

print("Top Customer:", top_customer_by_spend(orders, products, customers))
```

```python
def apply_offer(price, *discounts):

    for discount in discounts:
        price = price - (price * discount / 100)

    return price

print(apply_offer(1000, 10, 5))
```
