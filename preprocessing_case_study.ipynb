Handle missing values: For numerical columns,for categorical columns



```python
house['Sale Price']=house['Sale Price'].fillna(house['Sale Price'].median())
house['No of Bathrooms']=house['No of Bathrooms'].fillna(house['No of Bathrooms'].median())
house['Flat Area (in Sqft)']=house['Flat Area (in Sqft)'].fillna(house['Flat Area (in Sqft)'].median())
house['Lot Area (in Sqft)']=house['Lot Area (in Sqft)'].fillna(house['Lot Area (in Sqft)'].median())
house['Area of the House from Basement (in Sqft)']=house['Area of the House from Basement (in Sqft)'].fillna(house['Area of the House from Basement (in Sqft)'].median())
house['Zipcode']=house['Zipcode'].fillna(house['Zipcode'].median())
house['Latitude']=house['Latitude'].fillna(house['Latitude'].median())
house['Longitude']=house['Longitude'].fillna(house['Longitude'].median())
house['Living Area after Renovation (in Sqft)']=house['Living Area after Renovation (in Sqft)'].fillna(house['Living Area after Renovation (in Sqft)'].median())



house['No of Times Visited']=house['No of Times Visited'].fillna(house['No of Times Visited'].mode()[0])
house.isnull().sum()
```

```python
columns=['Sale Price','Flat Area (in Sqft)','Lot Area (in Sqft)','Area of the House from Basement (in Sqft)','Basement Area (in Sqft)',
    'Living Area after Renovation (in Sqft)']
```

```python
for i in columns:
  plt.figure(figsize=(5,5))
  plt.boxplot(house[i])
  plt.title(i)
```

```python
for col in columns:
  Q1=house[col].quantile(0.25)
  Q3=house[col].quantile(0.75)
  IQR=Q3-Q1
  lowerbound = Q1 - 1.5 * IQR
  upperbound = Q3 + 1.5 * IQR
print(Q1,Q3,IQR,lowerbound,upperbound)
```

```python
for x in house[col]:
  if x<lowerbound or x>upperbound:#outlier isidentified and print
    print(x)
```

```python
# 1. See where we started
print("Rows before removing outliers:", len(house))

# 2. Loop through ALL 6 columns
for col in columns:
    Q1 = house[col].quantile(0.25)
    Q3 = house[col].quantile(0.75)
    IQR = Q3 - Q1

    lowerbound = Q1 - 1.5 * IQR
    upperbound = Q3 + 1.5 * IQR

    # 3. OVERWRITE the dataset! Keep ONLY the rows that are inside the fences.
    house = house[(house[col] >= lowerbound) & (house[col] <= upperbound)]

# 4. See the final, clean dataset size
print("Rows after removing outliers:", len(house))
```

```python
for col in columns:
  house[col]=house[col].clip(lower=lowerbound,upper=upperbound)
  plt.boxplot(house[col])
  plt.title("boxplot")
  plt.show()
```
